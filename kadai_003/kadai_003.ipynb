```python
import pandas as pd
import openpyxl
from openpyxl.styles import PatternFill

# 処理対象のファイル名のリストを格納
file_list = ['2022_年間売上表.xlsx', '2023_年間売上表.xlsx']

# 結合後のデータを格納するための空のデータフレームを作成
df = pd.DataFrame()

# ファイルを一つずつ処理する
for file_name in file_list:
    # Excelファイルを読み込む
    temp_df = pd.read_excel(file_name)

    # データフレームの列名を指定
    temp_df.columns = ['売上年', '商品', '金額（千円）']

    # concat()でデータフレームを連結
    df = pd.concat([df, temp_df], ignore_index=True);

#グループ化
result = df.groupby(['商品', '売上年'])['金額（千円）'].sum().reset_index()

# Excelファイルに書き込む
result.to_excel('売上集計表.xlsx', sheet_name='Sheet1', index=False)

# Excelファイルを読み込む
workbook = openpyxl.load_workbook('売上集計表.xlsx')
# セルの色を（#F2F2F2）に設定
sheet = workbook.active

# A1, B1, C1セルの色を設定 (ARGB形式に修正)
for cell_name in ['A1', 'B1', 'C1']:
    sheet[cell_name].fill = PatternFill(patternType='solid', fgColor='FFF2F2F2')

workbook.save('売上集計表.xlsx')
```

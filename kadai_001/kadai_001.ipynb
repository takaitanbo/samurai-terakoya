```python
import openpyxl
from openpyxl.styles import Border, Side
from datetime import datetime

# 1. ワークブックとシートの作成
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "請求書"

# 2. データの入力
# 基本情報
today_formatted = datetime.now().strftime('%Y/%m/%d')

ws['B2'] = "請求書"
ws['B4'] = "株式会社ABC"
ws['B5'] = "〒101-0022 東京都千代田区神田練塀町300"
ws['B6'] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws['B7'] = "担当者名:鈴木一郎 様"

ws['F4'] = "No."
ws['G4'] = "0001"
ws['F5'] = "日付"
ws['G5'] = today_formatted

# 明細ヘッダー
headers = ["商品名", "数量", "単価", "金額"]
for col, text in enumerate(headers, start=2): # B列から
    ws.cell(row=10, column=col).value = text

# 明細データ
data = [
    ["商品A", 2, 10000, 20000],
    ["商品B", 1, 15000, 15000],
]

for r_idx, row_data in enumerate(data, start=11):
    for c_idx, value in enumerate(row_data, start=2):
        ws.cell(row=r_idx, column=c_idx).value = value

# 集計部分
ws['E13'] = 35000
ws['B15'] = "小計"
ws['E15'] = 35000
ws['B16'] = "消費税"
ws['E16'] = 3500 / 10 * 10 # 3500
ws['B17'] = "合計"
ws['E17'] = 38500

# 4. ファイル保存
today = datetime.now().strftime('%Y%m%d')
file_name = f"請求書_{today}.xlsx"
wb.save(file_name)
```
